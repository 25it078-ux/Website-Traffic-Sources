```python
# Website Traffic Sources Analysis

import pandas as pd
import matplotlib.pyplot as plt

# Sample Website Traffic Data
data = {
    'Traffic Source': [
        'Direct', 'Search Engine', 'Social Media', 'Referral', 'Email',
        'Direct', 'Search Engine', 'Social Media', 'Referral', 'Direct',
        'Search Engine', 'Email', 'Social Media', 'Referral', 'Direct',
        'Search Engine', 'Social Media', 'Email', 'Referral', 'Direct'
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("Website Traffic Dataset:")
print(df)

# Count Traffic Sources
traffic_counts = df['Traffic Source'].value_counts()

print("\nTraffic Source Counts:")
print(traffic_counts)

# Plot Bar Chart
plt.figure(figsize=(8,5))
traffic_counts.plot(kind='bar')
plt.title('Website Traffic Sources')
plt.xlabel('Traffic Source')
plt.ylabel('Number of Visitors')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Plot Pie Chart
plt.figure(figsize=(7,7))
traffic_counts.plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Website Traffic Sources Distribution')
plt.ylabel('')
plt.show()
```
